```sql
CREATE TABLE IF NOT EXISTS RPT_NHS_D01_KPI_BRANCH (
    -- Thông tin thời gian
    CDR_DT DATE COMMENT 'Ngày báo cáo (Định dạng: YYYY-MM-DD)',
    
    -- Thông tin phân loại và định mức
    ITM STRING COMMENT 'Phân loại cấp 1',
    KPI INT COMMENT 'KPI TKDP đã giao',
    
    -- Thông tin nhân viên (RM)
    RM_CODE STRING COMMENT 'Mã RM mở app',
    RM_NM STRING COMMENT 'Tên RM mở app',
    
    -- Thông tin tổ chức (Branch/Region)
    SUB_BRANCH_CODE STRING COMMENT 'Mã Chi nhánh mở app cấp 2',
    SUB_BRANCH_NAME STRING COMMENT 'Tên CN mở app cấp 2',
    BRANCH_CODE STRING COMMENT 'Mã Chi nhánh mở app cấp 1',
    BRANCH_NAME STRING COMMENT 'Tên CN mở app cấp 1',
    SUB_RGON STRING COMMENT 'Tiểu vùng của CN mở app',
    PRN_OU_TP STRING COMMENT 'Loại CN mở app',
    
    -- Kênh và chỉ số phát triển App
    CNL STRING COMMENT 'Kênh onboard app của KH',
    NBR_APP_NEW_D INT COMMENT 'SLKH mở mới app tại ngày báo cáo (có 1st login app = pDate)',
    NBR_APP_NEW_MTD INT COMMENT 'SLKH mở mới app tháng báo cáo',
    NBR_APP_NEW_YTD INT COMMENT 'SLKH mở mới app năm báo cáo',
    
    -- Chỉ số KPI thực đạt
    APP_KPI_MTD INT COMMENT 'SLKH mở mới app trong tháng báo cáo & đạt điều kiện KPI TKDP tại ngày',
    APP_KPI_YTD INT COMMENT 'SLKH mở mới app trong năm báo cáo & đạt điều kiện KPI TKDP tại ngày',
    
    -- Chỉ số kỹ thuật
    NBR_NFC_SCS INT COMMENT 'SLKH NFC thành công'
)
USING DELTA
PARTITIONED BY (CDR_DT)
COMMENT 'Báo cáo KPI phát triển App chi tiết theo nhân viên và chi nhánh, bao gồm định mức giao và kết quả thực hiện';
```
