```python
from pyspark.sql import functions as F
from pyspark.sql.functions import *

#df_src = spark.read.table("mb_poc.data_raw.app_smy")
df_src = (
    spark.read
    .option("header", "true")
    .option("encoding", "UTF-8")
    .option("inferSchema", "false")
    .csv("/Volumes/mb_poc/data_raw/uc2/RPT_NHS_D01_SMY.csv")
)

df_tgt = (
    df_src
    .filter("FLAG_NEW_CODE = 0")
    .select(
        col("CDR_DT").alias("CDR_DT"),
        col("FLAG_NEW_CODE").alias("FLAG_NEW_CODE"),
        col("NEW_USR_GRP").alias("NEW_USR_GRP"),
        col("APP_NOT_LOGIN_GRP").alias("APP_NOT_LOGIN_GRP"),
        col("MKT_CNL_GRP").alias("MKT_CNL_GRP"),
        col("AGE_GRP").alias("AGE_GRP"),
        F.regexp_replace(col("NBR_APP"), ",", "").cast("int").alias("NBR_APP"),
        F.regexp_replace(col("NBR_APP_NEW_D"), ",", "").cast("int").alias("NBR_APP_NEW_D"),
        F.regexp_replace(col("NBR_APP_NEW_MTD"), ",", "").cast("int").alias("NBR_APP_NEW_MTD"),
        F.regexp_replace(col("NBR_APP_NEW_YTD"), ",", "").cast("int").alias("NBR_APP_NEW_YTD"),
        F.regexp_replace(col("NBR_APP_NEW_LM"), ",", "").cast("int").alias("NBR_APP_NEW_LM"),
        F.regexp_replace(col("NBR_ACTV_APP"), ",", "").cast("int").alias("NBR_ACTV_APP"),
        F.regexp_replace(col("APP_MTD_ACTV"), ",", "").cast("int").alias("APP_MTD_ACTV"),
        F.regexp_replace(col("NBR_REACTV_LM"), ",", "").cast("int").alias("NBR_REACTV_LM"),
        F.regexp_replace(col("NBR_INACTV_LM"), ",", "").cast("int").alias("NBR_INACTV_LM"),
        F.regexp_replace(col("NBR_REACTV_LD"), ",", "").cast("int").alias("NBR_REACTV_LD"),
        F.regexp_replace(col("NBR_INACTV_LD"), ",", "").cast("int").alias("NBR_INACTV_LD"),
        F.regexp_replace(col("NBR_ACTV_LOGIN"), ",", "").cast("int").alias("NBR_ACTV_LOGIN"),
        F.regexp_replace(col("APP_NEW_ACTV_LOGIN"), ",", "").cast("int").alias("APP_NEW_ACTV_LOGIN"),
        F.regexp_replace(col("NBR_ACTV_TXN_3M"), ",", "").cast("int").alias("NBR_ACTV_TXN_3M"),
        F.regexp_replace(col("NBR_ACTV_TXN_1M"), ",", "").cast("int").alias("NBR_ACTV_TXN_1M"),
        F.regexp_replace(col("APP_NEW_ACTV_TXN_3M"), ",", "").cast("int").alias("APP_NEW_ACTV_TXN_3M"),
        F.regexp_replace(col("NBR_NFC_SCS"), ",", "").cast("int").alias("NBR_NFC_SCS"),
        F.regexp_replace(col("NBR_CHURN_IN_D"), ",", "").cast("int").alias("NBR_CHURN_IN_D")
    )
)

# df_tgt.write.format("delta") \
#     .mode("overwrite") \
#     .saveAsTable("mb_poc.gold.rpt_nhs_d01_smy")
df_tgt.write.mode("append").insertInto("mb_poc.gold.rpt_nhs_d01_smy")
```
