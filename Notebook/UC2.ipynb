```python
# Đọc dữ liệu từ file CSV
df = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load("/Volumes/mb_poc/data_raw/uc2/KHCN_HDV_D09_DGTAL_SVG.csv")

# Hiển thị một vài dòng dữ liệu để kiểm tra
print(f"Số dòng: {df.count()}")
print(f"Số cột: {len(df.columns)}")
display(df.limit(5))

# Ghi dữ liệu vào bảng Delta
df.write.format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("feautures")

print("✓ Đã ghi dữ liệu thành công vào bảng feautures")
```
