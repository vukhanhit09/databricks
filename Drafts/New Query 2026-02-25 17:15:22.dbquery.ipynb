```sql
WITH tong_toan_he AS (
  SELECT 
    SUM(CLS_BOOK_BAL) AS tong_so_du_toan_he,
    SUM(NBR_OF_AC_W_BAL_TDY) AS tong_so_toan_he,
    SUM(NBR_OF_TVR_CST_TDY) AS tong_kh_toan_he
  FROM mb_poc.gold.rpt_khcn_hdv_d09_dgtal_svg
  WHERE PD_TP = 'TGTK'
    AND CDR_DT = (SELECT MAX(CDR_DT) FROM mb_poc.gold.rpt_khcn_hdv_d09_dgtal_svg)
)

SELECT 
  'TỔNG TOÀN HỆ THỐNG' AS Cap_do,
  'Tất cả' AS Phan_loai,
  tong_so_du_toan_he AS So_du_tiet_kiem,
  100.00 AS Ty_trong_phan_tram,
  tong_so_toan_he AS So_luong_so,
  tong_kh_toan_he AS So_luong_khach_hang,
  ROUND(tong_so_du_toan_he / NULLIF(tong_so_toan_he, 0), 0) AS So_du_binh_quan_tren_so,
  ROUND(tong_so_du_toan_he / NULLIF(tong_kh_toan_he, 0), 0) AS So_du_binh_quan_tren_KH
FROM tong_toan_he

UNION ALL

SELECT 
  'Theo Khối' AS Cap_do,
  CST_LOB_NM AS Phan_loai,
  SUM(CLS_BOOK_BAL) AS So_du_tiet_kiem,
  ROUND(SUM(CLS_BOOK_BAL) * 100.0 / (SELECT tong_so_du_toan_he FROM tong_toan_he), 2) AS Ty_trong_phan_tram,
  SUM(NBR_OF_AC_W_BAL_TDY) AS So_luong_so,
  SUM(NBR_OF_TVR_CST_TDY) AS So_luong_khach_hang,
  ROUND(SUM(CLS_BOOK_BAL) / NULLIF(SUM(NBR_OF_AC_W_BAL_TDY), 0), 0) AS So_du_binh_quan_tren_so,
  ROUND(SUM(CLS_BOOK_BAL) / NULLIF(SUM(NBR_OF_TVR_CST_TDY), 0), 0) AS So_du_binh_quan_tren_KH
FROM mb_poc.gold.rpt_khcn_hdv_d09_dgtal_svg, tong_toan_he
WHERE PD_TP = 'TGTK'
  AND CDR_DT = (SELECT MAX(CDR_DT) FROM mb_poc.gold.rpt_khcn_hdv_d09_dgtal_svg)
GROUP BY CST_LOB_NM

UNION ALL

SELECT 
  'Theo Phân khúc' AS Cap_do,
  CST_GRP AS Phan_loai,
  SUM(CLS_BOOK_BAL) AS So_du_tiet_kiem,
  ROUND(SUM(CLS_BOOK_BAL) * 100.0 / (SELECT tong_so_du_toan_he FROM tong_toan_he), 2) AS Ty_trong_phan_tram,
  SUM(NBR_OF_AC_W_BAL_TDY) AS So_luong_so,
  SUM(NBR_OF_TVR_CST_TDY) AS So_luong_khach_hang,
  ROUND(SUM(CLS_BOOK_BAL) / NULLIF(SUM(NBR_OF_AC_W_BAL_TDY), 0), 0) AS So_du_binh_quan_tren_so,
  ROUND(SUM(CLS_BOOK_BAL) / NULLIF(SUM(NBR_OF_TVR_CST_TDY), 0), 0) AS So_du_binh_quan_tren_KH
FROM mb_poc.gold.rpt_khcn_hdv_d09_dgtal_svg, tong_toan_he
WHERE PD_TP = 'TGTK'
  AND CDR_DT = (SELECT MAX(CDR_DT) FROM mb_poc.gold.rpt_khcn_hdv_d09_dgtal_svg)
GROUP BY CST_GRP

UNION ALL

SELECT 
  'Theo Vùng' AS Cap_do,
  SUB_RGON AS Phan_loai,
  SUM(CLS_BOOK_BAL) AS So_du_tiet_kiem,
  ROUND(SUM(CLS_BOOK_BAL) * 100.0 / (SELECT tong_so_du_toan_he FROM tong_toan_he), 2) AS Ty_trong_phan_tram,
  SUM(NBR_OF_AC_W_BAL_TDY) AS So_luong_so,
  SUM(NBR_OF_TVR_CST_TDY) AS So_luong_khach_hang,
  ROUND(SUM(CLS_BOOK_BAL) / NULLIF(SUM(NBR_OF_AC_W_BAL_TDY), 0), 0) AS So_du_binh_quan_tren_so,
  ROUND(SUM(CLS_BOOK_BAL) / NULLIF(SUM(NBR_OF_TVR_CST_TDY), 0), 0) AS So_du_binh_quan_tren_KH
FROM mb_poc.gold.rpt_khcn_hdv_d09_dgtal_svg, tong_toan_he
WHERE PD_TP = 'TGTK'
  AND CDR_DT = (SELECT MAX(CDR_DT) FROM mb_poc.gold.rpt_khcn_hdv_d09_dgtal_svg)
GROUP BY SUB_RGON

UNION ALL

SELECT 
  'Theo Chi nhánh' AS Cap_do,
  BRANCH_NAME AS Phan_loai,
  SUM(CLS_BOOK_BAL) AS So_du_tiet_kiem,
  ROUND(SUM(CLS_BOOK_BAL) * 100.0 / (SELECT tong_so_du_toan_he FROM tong_toan_he), 2) AS Ty_trong_phan_tram,
  SUM(NBR_OF_AC_W_BAL_TDY) AS So_luong_so,
  SUM(NBR_OF_TVR_CST_TDY) AS So_luong_khach_hang,
  ROUND(SUM(CLS_BOOK_BAL) / NULLIF(SUM(NBR_OF_AC_W_BAL_TDY), 0), 0) AS So_du_binh_quan_tren_so,
  ROUND(SUM(CLS_BOOK_BAL) / NULLIF(SUM(NBR_OF_TVR_CST_TDY), 0), 0) AS So_du_binh_quan_tren_KH
FROM mb_poc.gold.rpt_khcn_hdv_d09_dgtal_svg, tong_toan_he
WHERE PD_TP = 'TGTK'
  AND CDR_DT = (SELECT MAX(CDR_DT) FROM mb_poc.gold.rpt_khcn_hdv_d09_dgtal_svg)
GROUP BY BRANCH_NAME

ORDER BY 
  CASE Cap_do
    WHEN 'TỔNG TOÀN HỆ THỐNG' THEN 1
    WHEN 'Theo Khối' THEN 2
    WHEN 'Theo Phân khúc' THEN 3
    WHEN 'Theo Vùng' THEN 4
    WHEN 'Theo Chi nhánh' THEN 5
  END,
  So_du_tiet_kiem DESC
```
